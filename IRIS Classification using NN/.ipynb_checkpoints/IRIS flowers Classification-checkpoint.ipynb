```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train_tensor=torch.tensor(x_train,dtype=torch.float)
x_test_tensor=torch.tensor(x_test,dtype=torch.int)
y_train_tensor=torch.tensor(y_train,dtype=torch.float)
y_test_tensor=torch.tensor(y_test,dtype=torch.int)
```

```python
train_dataset=TensorDataset(x_train_tensor,y_train_tensor)
test_dataset=TensorDataset(x_test_tensor,y_test_tensor)
```

```python
train_loader=DataLoader(train_dataset,shuffle=True)
test_loader=DataLoader(test_dataset,shuffle=False)
```

```python
class IrisModel(nn.Module):
    def __init__(self,in_features,out_features):
        super().__init__()
        self.fc1=nn.Linear(in_features=in_features,out_features=32)
        self.fc2=nn.Linear(in_features=32,out_features=64)
        self.out=nn.Linear(in_features=64,out_features=3)
        self.relu=nn.ReLU()
        self.softmax=nn.Softmax()

    def forward(self,x):
        x=self.fc1(x)
        x=self.relu(x)
        x=self.fc2(x)
        x=self.relu(x)
        x=self.out(x)
        x=self.softmax(x)

        return x
```

```python
lr=0.001
epochs=500
```

```python
model=IrisModel(in_features=4,out_features=3)
```

```python
optimizer=optim.Adam(model.parameters(),lr=lr)
loss_fn=nn.CrossEntropyLoss()
```

```python
train_losses=[]
test_losses=[]

for epoch in range(epochs):
    running_loss=0.0
    running_test_loss=0.0
    model.train()
    for x_batch,y_batch in train_loader:
        y_batch = y_batch.view(-1).long()
        model.zero_grad()
        pred=model(x_batch)
        loss=loss_fn(pred,y_batch)
        running_loss+=loss.item()
        loss.backward()
        optimizer.step()
    train_losses.append(running_loss/len(train_loader))
    model.eval()

    with torch.no_grad():
        for x_batch,y_batch in test_loader:
            x_batch = x_batch.float()
            y_batch=y_batch.view(-1).long()
            pred=model(x_batch)
            loss=loss_fn(pred,y_batch)
            running_test_loss += loss.item()
        test_losses.append(
            running_test_loss / len(test_loader)
        )
        if epoch % 100==0:
            print(
            f"Epoch {epoch+1}/{epochs}, "
            f"Train Loss: {train_losses[-1]:.4f}, "
            f"Test Loss: {test_losses[-1]:.4f}"
        )
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(train_losses,label="Training Loss")
plt.plot(test_losses,label="Test Loss")
plt.title("Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()
```
