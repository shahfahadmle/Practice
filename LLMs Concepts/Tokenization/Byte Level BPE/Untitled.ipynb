```python
import tiktoken
import pandas as pd
```

```python
tokenizer=tiktoken.get_encoding('gpt2')
```

```python
# Encode text into token IDs and decode back
tokens = tokenizer.encode("Hello, world!")
print(tokens)  # [13225, 11, 2375, 0]
print(tokenizer.decode(tokens))  # "Hello, world!"
```

```python
file_path = "Dataset/input.txt"

# Open and read the entire file into a single string
with open(file_path, "r", encoding="utf-8") as file:
    raw_text = file.read()

print(f"Loaded {len(raw_text)} characters.")
```

```python
tokens=tokenizer.encode(raw_text)
```

```python
print(tokens[:100])
```

```python
tokenizer.decode(tokens[:100])
```
