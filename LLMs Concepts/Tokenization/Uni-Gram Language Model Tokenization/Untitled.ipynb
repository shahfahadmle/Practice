```python
import sentencepiece as spm
```

```python
spm.SentencePieceTrainer.train(
    input='Dataset/input.txt',
    model_prefix='unigram_model',
    model_type='unigram',
    vocab_size=8000,
    character_coverage=0.9995,
    unk_id=0,
    pad_id=1,
    bos_id=2,
    eos_id=3,
)
```

```python
tokenizer=spm.SentencePieceProcessor(
    model_file='unigram_model.model'
)
```

```python
text=' i love machine learning'
tokens=tokenizer.encode(
    text,
    out_type=str
)
print(tokens)
```

```python
ids=tokenizer.encode(
    text,
    out_type=int
)
print(ids)
```

```python
decoded = tokenizer.decode(ids)

print(decoded)
```

```python
tokens = tokenizer.sample_encode_as_pieces(
    "playing machine learning",
    nbest_size=4,
    alpha=0.1
)

print(tokens)
```
