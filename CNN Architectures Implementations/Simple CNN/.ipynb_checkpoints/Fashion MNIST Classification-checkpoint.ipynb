```python
import pandas as pd
import numpy as np
from torch.utils.data import Dataset,DataLoader
import torch 
import matplotlib.pyplot as plt 
from PIL import Image
from pathlib import Path
import torchvision.transforms as transforms
import os
```

```python
class_names = {
    0: "T-shirt/top",
    1: "Trouser",
    2: "Pullover",
    3: "Dress",
    4: "Coat",
    5: "Sandal",
    6: "Shirt",
    7: "Sneaker",
    8: "Bag",
    9: "Ankle boot"
}
```

```python
class FashionMNISTDataset(Dataset):
    def __init__(self,data,transform=None):
        self.data=data
        self.transform=transform
        self.labels=torch.tensor(data.iloc[:,0].values,dtype=torch.long)
        features=data.iloc[:,1:].values
        features=features.reshape(-1,1,28,28)/255.0
        self.features=torch.tensor(features,dtype=torch.float32)


    def __len__(self):
        return len(self.labels)

    def __getitem__(self,idx):
        x=self.features[idx]
        y=self.labels[idx]

        if self.transform:
            x=self.transform(x)
        return x,y
        
```

```python
train_data=pd.read_csv('Dataset/fashion-mnist_train.csv')
test_data=pd.read_csv('Dataset/fashion-mnist_test.csv')
```

```python
train_dataset=FashionMNISTDataset(train_data)
test_dataset=FashionMNISTDataset(test_data)
```

```python
train_loader=DataLoader(train_dataset,
                        batch_size=32,
                        shuffle=True
                       )
test_loader=DataLoader(test_dataset,
                       batch_size=32,
                       shuffle=False
                      )
```

```python
from cnn_modle import CNN
```

```python
model=CNN(num_classes=10)
```

```python
import torch.optim as optim
import torch.nn as nn
from tqdm.auto import tqdm
```

```python

optimizer=optim.Adam(model.parameters(),lr=0.001)
loss_fn=nn.CrossEntropyLoss()
```

```python
epochs=10
train_loss_history=[]
test_loss_history=[]
for epoch in tqdm(range(epochs),desc="Training Epochs"):
    running_loss=0.0
    model.train()
    progress_bar = tqdm(
        train_loader,
        desc=f"Epoch {epoch+1}/{epochs}",
        leave=False
    )
    for image,label in train_loader:
        pred=model(image)
        loss=loss_fn(pred,label)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        running_loss+=loss.item()
        progress_bar.set_postfix(loss=f"{loss.item():.4f}")
    epoch_loss=running_loss/len(train_loader)
    train_loss_history.append(epoch_loss)
    if epoch % 100 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {epoch_loss:.4f}")
    
```

```python
model.eval()
test_loss=0.0
correct=0
total=0
with torch.no_grad():
    for image ,labels in test_loader:
        outputs=model(image)
        loss=loss_fn(outputs,labels)
        test_loss+=loss.item()*image.size(0)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
    avg_test_loss = test_loss / total
    test_accuracy = (correct / total) * 100
    
    print(f"Test Loss: {avg_test_loss:.4f} | Test Accuracy: {test_accuracy:.2f}%")
        
    
```

```python
torch.save(model.state_dict(),"Samall_fashionMNIST_model.pth")
```

```python
def predict_new_image(model,image_path):
    image=Image.open(image_path).convert('L')
    transform=transforms.Compose([
        transforms.Resize((28,28)),
        transforms.ToTensor(),
        transforms.Normalize( (0.5,),(0.5)),
        
    ])
    image_tensor=transform(image)
    image_tensor=image_tensor.unsqueeze(0)
    model.eval()
    with torch.no_grad():
        pred=model(image_tensor)
        predicted_class=torch.argmax(pred).item()

    return predicted_class
        
```

```python
new_model=CNN()
state_dict=torch.load('Samall_fashionMNIST_model.pth',weights_only=True)
new_model.load_state_dict(state_dict)
```

```python
image_path1 = r"Testing\test_image.jfif"

pred1 = predict_new_image(new_model, image_path1)

print("Predicted Class:", class_names.get(pred1))
```
